```python

# Assignment 3.1: Histogram Generation

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_name = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imdecode(
    np.frombuffer(uploaded[image_name], np.uint8),
    cv2.IMREAD_GRAYSCALE
)

print("Image uploaded successfully!")
print("Image shape:", img.shape)
```

```python

# Generate histogram

histogram = cv2.calcHist(
    [img], [0], None, [256], [0, 256]
)

# Display original image and histogram
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.plot(histogram, color="black")
plt.title("Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.xlim([0, 256])

plt.tight_layout()
plt.show()

print("Histogram generated successfully!")
```

```python

# Assignment 3.2: Normalised Histogram (PDF)

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_name = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[image_name], np.uint8),
    cv2.IMREAD_GRAYSCALE
)

print("Image uploaded successfully!")
```

```python

# Calculate histogram
hist = cv2.calcHist([img], [0], None, [256], [0, 256])

# Normalize histogram to get PDF
total_pixels = img.size
pdf = hist.flatten() / total_pixels

# Display original image and normalized histogram
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.plot(pdf, color="blue")
plt.title("Normalized Histogram (PDF)")
plt.xlabel("Pixel Intensity")
plt.ylabel("Probability")
plt.xlim([0, 256])

plt.tight_layout()
plt.show()

print("Sum of PDF:", np.sum(pdf))
print("Normalized Histogram generated successfully!")
```

```python

# Assignment 3.3: Histogram Equalization (CDF)

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_name = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[image_name], np.uint8),
    cv2.IMREAD_GRAYSCALE
)

print("Image uploaded successfully!")
```

```python

# Calculate histogram
hist = cv2.calcHist([img], [0], None, [256], [0, 256]).flatten()

# Calculate normalized histogram (PDF)
pdf = hist / img.size

# Calculate cumulative distribution function (CDF)
cdf = np.cumsum(pdf)

# Create transformation lookup table
transform = np.round(cdf * 255).astype(np.uint8)

# Apply CDF transformation
equalized = transform[img]

# Display original and equalized images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("CDF Histogram Equalization")
plt.axis("off")

plt.tight_layout()
plt.show()

print("CDF Histogram Equalization completed!")
```

```python

# Display CDF graph

plt.figure(figsize=(8, 4))

plt.plot(cdf, color="red")
plt.title("Cumulative Distribution Function (CDF)")
plt.xlabel("Pixel Intensity")
plt.ylabel("Cumulative Probability")
plt.xlim([0, 255])
plt.ylim([0, 1.05])
plt.grid()

plt.show()
```

```python

# Assignment 3.4: Histogram Equalization

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_name = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[image_name], np.uint8),
    cv2.IMREAD_GRAYSCALE
)

print("Image uploaded successfully!")
```

```python

# Apply Histogram Equalization
equalized = cv2.equalizeHist(img)

# Display original and equalized images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("Equalized Image")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Histogram Equalization completed!")
```

```python

# Calculate original and equalized histograms

original_hist = cv2.calcHist(
    [img], [0], None, [256], [0, 256]
)

equalized_hist = cv2.calcHist(
    [equalized], [0], None, [256], [0, 256]
)

# Display histogram comparison
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(original_hist, color="blue")
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")

plt.subplot(1, 2, 2)
plt.plot(equalized_hist, color="red")
plt.title("Equalized Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")

plt.tight_layout()
plt.show()

print("Histogram comparison completed!")
```
