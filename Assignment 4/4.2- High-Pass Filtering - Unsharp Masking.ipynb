```python

# Assignment 4.1: Spatial Filtering - Correlation

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_name = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imdecode(
    np.frombuffer(uploaded[image_name], np.uint8),
    cv2.IMREAD_GRAYSCALE
)

print("Image uploaded successfully!")
print("Image shape:", img.shape)

# Display original image
plt.figure(figsize=(6, 5))
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

```python

# Define 3x3 averaging kernel
kernel = np.ones((3, 3), np.float32) / 9

# Apply correlation using OpenCV
correlation = cv2.filter2D(
    img, cv2.CV_32F, kernel
)

# Convert output to uint8
correlation_img = np.clip(
    correlation, 0, 255
).astype(np.uint8)

# Display results
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(correlation_img, cmap="gray")
plt.title("Correlation Output")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Correlation filtering completed!")
```

```python

# Sobel horizontal edge detection kernel
sobel_kernel = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float32)

# Apply correlation
edge_output = cv2.filter2D(
    img, cv2.CV_32F, sobel_kernel
)

# Display edge magnitude
edge_magnitude = np.abs(edge_output)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edge_magnitude, cmap="gray")
plt.title("Edge Detection Using Correlation")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Edge detection completed!")
```

```python

# Assignment 4.2: Unsharp Masking

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_name = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imdecode(
    np.frombuffer(uploaded[image_name], np.uint8),
    cv2.IMREAD_GRAYSCALE
)

print("Image uploaded successfully!")
print("Image shape:", img.shape)

# Display original image
plt.figure(figsize=(6, 5))
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python

# Apply Gaussian Blur
blurred = cv2.GaussianBlur(
    img, (5, 5), 1.0
)

# Display original and blurred images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(blurred, cmap="gray")
plt.title("Gaussian Blurred Image")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Gaussian Blur completed!")
```

```python

# Convert images to float
original = img.astype(np.float32)
blur = blurred.astype(np.float32)

# Create unsharp mask
mask = original - blur

# Set sharpening strength
amount = 1.5

# Apply unsharp masking
sharpened = original + amount * mask

# Clip pixel values
sharpened = np.clip(
    sharpened, 0, 255
).astype(np.uint8)

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(np.clip(np.abs(mask) * 3, 0, 255), cmap="gray")
plt.title("Unsharp Mask")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharpened, cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Unsharp Masking completed!")
```

```python

# Compare original and sharpened images

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sharpened, cmap="gray")
plt.title("Unsharp Masking Result")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Image sharpening comparison completed!")
```
